```python
#Import
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import mean_absolute_error
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

import seaborn as sns
import matplotlib.pyplot as plt

#Load data
df = pd.read_csv('./edinet_growth_analysis/output/growth_rate_overEqual40.csv')
df_plus = pd.read_csv('./growth_40_whole.csv')

print(df)
```

```python
name_mapping = {
    '本田技研工業': 'Honda',
    'バンダイナムコホールディングス': 'Bandai',
    '日産自動車': 'Nissan',
    'トヨタ自動車': 'Toyota',
    'マツダ': 'Mazda'
}

df['company'] = df['company'].replace(name_mapping)
df_plus['company'] = df_plus['company'].replace(name_mapping)
```

```python
print(df_plus)
```

```python
# Bar plot for each company (2022 to 2026)
plt.figure(figsize=(10, 6))
sns.barplot(x='company', y='growth_rate_2022to2026', data=df, palette='viridis')

plt.title('Growth rate from 2022 to 2026')
plt.ylabel('Growth rate (%)')
plt.xlabel('Companies')
plt.show()
```

```python
year_mapping = {
    'growth_rate_duration1': 'prior1year',
    'growth_rate_duration2': 'prior2years',
    'growth_rate_duration3': 'prior3years',
    'growth_rate_duration4': 'prior4years'
}

df_clean = df_plus.rename(columns=year_mapping)

print(df_clean)
```

```python
year_mapping = {
    'growth_rate_duration1': '2025→2026',
    'growth_rate_duration2': '2024→2026',
    'growth_rate_duration3': '2023→2026',
    'growth_rate_duration4': '2022→2026'
}

df_clean = df_plus.rename(columns=year_mapping)

print(df_clean)
```

```python
df_years = df_clean.set_index('company')[['2022→2026','2023→2026', '2024→2026', '2025→2026']]

df_flipped = df_years.T

plt.figure(figsize=(10, 6))

sns.lineplot(data=df_flipped, markers=True, dashes=False, linewidth=2.5)

plt.axhline(y=40, color='grey', linestyle='--', linewidth=1.5, label='40% Threshold')
plt.title('Growth Rate Trend (2022 - 2026)')
plt.ylabel('Growth Rate (%)')
plt.xlabel('Duration')
plt.grid(True)
plt.legend(title='Companies')
plt.show()
```

```python
def avg_growth_rate(df_name,row):
    result = 0
    for rate in df_name[row]:
        result += rate
    return round(result / len(df_name[row]), 2)

avg_rate_4 = avg_growth_rate(df_clean, 'prior4years')
avg_rate_3 = avg_growth_rate(df_clean, 'prior3years')
avg_rate_2 = avg_growth_rate(df_clean, 'prior2years')
avg_rate_1 = avg_growth_rate(df_clean, 'prior1year')

plt.figure(figsize=(8, 4))
sns.barplot(x=df_flipped.index, y=df_flipped['Toyota'], palette='viridis')

plt.title('Toyota Growth Rate Trend (2022 - 2026)')
plt.ylabel('Growth rate (%)')
plt.xlabel('Years')
plt.show()
```
